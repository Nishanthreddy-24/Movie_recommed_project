Loading data

```python
import pandas as pd
movies = pd.read_csv("tmdb_5000_movies.csv")
credits = pd.read_csv("tmdb_5000_credits.csv")
movies = movies.merge(credits, on="title")
movies.shape
```

Explore the data

```python
movies = movies[["movie_id", "title", "overview", "genres", "keywords", "cast", "crew"]]
movies.head()
```

Drop the missing values

```python
movies = movies.dropna()
movies = movies.drop_duplicates()
movies = movies.reset_index(drop=True)
len(movies)
```

```python
import ast
def names(text):
    return [d["name"] for d in ast.literal_eval(text)]
def director(text):
    return [d["name"] for d in ast.literal_eval(text) if d["job"] == "Director"]

movies["genres"] = movies["genres"].apply(names)
movies["keywords"] = movies["keywords"].apply(
    lambda x: names(x)[:3]
)
movies["crew"] = movies["crew"].apply(director)
movies.iloc[0]
```

Prepare the data

```python
movies["overview"] = movies["overview"].str.split()
for col in ["genres", "keywords", "cast", "crew"]:
    movies[col] = movies[col].apply(
        lambda x: [i.replace(" ", "") for i in x]
        )
movies["tags"] = (movies["overview"] + movies["genres"] + movies["keywords"] + movies["cast"] + movies["crew"])
movies.head()
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
movies["tags"] = movies["tags"].apply(lambda x: " ".join(x).lower())
movies["tags"] = movies["tags"].apply(lambda x: " ".join(ps.stem(w) for w in x.split()))
print(ps.stem("loving"), ps.stem("loved"))
```

Text to vectors

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words="english")
vectors = cv.fit_transform(movies["tags"]).toarray()
vectors.shape
```

Predict similar movies using cosine_similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
similarity.shape
```

```python
def recommend(movie):
    index = movies[movies["title"] == movie].index[0]
    scores = list(enumerate(similarity[index]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    for i, score in scores[1:6]:
        print(movies.iloc[i].title)
```

Result....!

```python
recommend("Avatar")
```
